#### **Multiplicación escalar de dos vectores**

Para la multiplicación escalar de vectores, se encontraron dos métodos por medio de la libreería numpy. Uno de ellos, se lleva a cabo con el operador @ y el otro, se realiza por medio de la función .dot

```python
#Creación de vectores
a= np.array([3.1,1, -0.5, -3.2, 6])
b=np.array([1, 3, 2.2, 5.1, 1])

#Multiplicación escalar de los vectores
mul=a@b
print("La multiplicación escalar de los vectores es:\n", mul)
```

```python
multi=np.dot(a,b)
print(multi)
```

#### **Multiplicación punto a punto**

```python
puntito = a * b
print(puntito)
```

#### **Función 𝑦[𝑛]=sin (𝜋∗0.12𝑛) en el intervalo 0 <= n <= 100**

```python
n = np.arange(0, 101)
y = np.sin(np.pi * 0.12 * n)
```

#### **Función 𝑦2[𝑛]=cos(2𝜋∗0.03𝑛)**

```python
y2 = np.cos(2 * np.pi * 0.03 * n)
```

#### **Suma de señales**

```python
s = y + y2
t = y * y2
```

#### **Gráfica señales**

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(n, y, label='y[n]', color='purple')
plt.plot(n, y2, label='y2[n]', color='turquoise')
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.title('Señales y[n] y y2[n]')
plt.legend()
plt.show()
```

#### **Gráfica suma de señales**

```python
plt.figure()
plt.plot(n, s, label='s[n]', color='deeppink')
plt.plot(n, t, label='t[n]', color='skyblue')
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.title('Señales s[n] y t[n]')
plt.legend()
plt.show()
```

### **Función notas**

```python
def estadisticas_notas(dic):
    notas = pd.Series(dic)
    return pd.Series({
        'min': notas.min(),
        'max': notas.max(),
        'media': notas.mean(),
        'desviacion': notas.std()
    })

# Ejemplo
notas = {'Ana': 4.5, 'Luis': 3.8, 'Pedro': 4.2}
print(estadisticas_notas(notas))
```

#### **Cálculo del IMC**

```python
df['BMI'] = df['weight'] / ((df['height'] / 100) ** 2)
```

```python
def clasificar_imc(imc):
    if imc < 18.5:
        return 'Bajo peso'
    elif imc < 25:
        return 'Normal'
    elif imc < 30:
        return 'Sobrepeso'
    else:
        return 'Obesidad'

df['Categoria'] = df['BMI'].apply(clasificar_imc)

print(df.head())
```
